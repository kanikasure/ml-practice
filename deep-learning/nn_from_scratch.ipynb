```python
from sklearn.datasets import make_moons
import numpy as np
```

```python
X, y = make_moons(n_samples = 500, noise = 0.2, random_state = 42)
```

```python
#initializing parameters - 17 trainable parameters
def init_params(n_in = 2, n_hidden = 4, n_out= 1):
  b1 = np.zeros(n_hidden) #biases for layer 1 nodes
  b2 = np.zeros(n_out) #bias for layer2 node = output node
  w1 = 0.1*np.random.randn(n_hidden, n_in) #weights for layer 1
  w2 = 0.1*np.random.randn(n_out, n_hidden) #weights for layer 2
  return {"w1": w1, "b1": b1, "w2": w2, "b2": b2}
```

```python
params =  init_params()
```

```python
def sigmoid(z):
  return 1/(1+ np.exp(-z))
```

```python
def forward(X, params):
  a0 = X #500x2

  a1 = np.tanh(np.dot(a0, params["w1"].T) + params["b1"]) #(500x2).(2x4) + (1x4)

  a2 = sigmoid(np.dot(a1, params["w2"].T) + params["b2"])

  return a1, a2
```

```python
a1, a2 = forward(X, params)
```
